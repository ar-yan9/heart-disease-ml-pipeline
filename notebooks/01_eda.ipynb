```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Working dataset URL (UCI Heart Disease via Kaggle mirror)
url = "https://raw.githubusercontent.com/sharmaroshan/Heart-UCI-Dataset/master/heart.csv"
df = pd.read_csv(url)
print(df.shape)
df.head()
```

```python
print("Missing values:\n", df.isnull().sum())
print("\nData types:\n", df.dtypes)
print("\nBasic stats:\n")
df.describe()
```

```python
print("Missing values:\n", df.isnull().sum())
print("\nData types:\n", df.dtypes)
print("\nBasic stats:\n")
df.describe()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='target', data=df, palette='Set2')
plt.title("Disease (1) vs No Disease (0)")
plt.savefig("target_distribution.png", bbox_inches='tight')  # save BEFORE show
plt.show()
print("✅ saved")
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.savefig("correlation_heatmap.png", bbox_inches='tight')
plt.show()
print("✅ saved")
```

```python
df.hist(figsize=(14,10), color='steelblue', edgecolor='black')
plt.suptitle("Feature Distributions")
plt.tight_layout()
plt.savefig("feature_distributions.png", bbox_inches='tight')
plt.show()
print("✅ saved")
```
